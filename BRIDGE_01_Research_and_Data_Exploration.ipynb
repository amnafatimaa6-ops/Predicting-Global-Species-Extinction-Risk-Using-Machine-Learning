```python
BASE = Path("/content/PROJECT_BRIDGE")

experiment_folder = BASE / "04_Experiments"
experiment_folder.mkdir(parents=True, exist_ok=True)

#  Research protocol
protocol = """
# PROJECT BRIDGE — Experimental Protocol

## Research objective
Compare human-only and AI-assisted workflows to investigate
whether AI creates measurable operational value for NOVA Retail.

## Experiment A: Customer Enquiry Management
Benchmark: 100 synthetic customer enquiries.

Human-only:
- Read the enquiry and fictional policy.
- Write the response independently.
- Record completion time.

AI-assisted:
- Provide the same enquiry and policy to the AI system.
- Record AI processing time.
- Review and correct the generated response.
- Record total elapsed time and human review time.

Evaluation:
- Policy accuracy.
- Completeness.
- Unsupported claims.
- Escalation correctness.
- Correction count.
- Time and estimated labour cost.

## Experiment B: Customer Feedback Analysis
Benchmark: 120 synthetic customer reviews.

Tasks:
- Sentiment classification.
- Feedback topic classification.

Model input:
- review_text only.

Targets:
- sentiment_label.
- feedback_topic.

Evaluation:
- Accuracy.
- Macro precision.
- Macro recall.
- Macro F1.
- Confusion matrix.
- Processing time.

Use a stratified train/test split.
Keep the test set separate from model training and model selection.

Important limitation:
The labels were manually constructed by one annotator.
Results measure agreement with this synthetic annotation scheme,
not independently verified real-world sentiment.

Do not claim an independent human accuracy comparison unless
a separate reviewer independently labels the test records.

## Experiment C: Product Description Generation
Benchmark: 30 synthetic product records.

Human-only:
- Write a product description using catalogue facts.

AI-assisted:
- Generate a description using the same catalogue facts.
- Review and correct the AI draft.

Evaluation:
- Factual accuracy.
- Completeness.
- Readability.
- Consistency.
- Unsupported claims.
- Human correction time.
- Total completion time.

## Fairness and consistency
- Both workflows receive the same source information.
- Use a fixed task brief and evaluation rubric.
- Record the AI model and version where available.
- Record prompts and model settings.
- Record AI processing time separately from human review time.
- Randomise or alternate task order where practical.
- Do not change evaluation criteria after seeing results.

## Financial evaluation
Calculate labour cost using measured human time and an explicitly
stated hourly labour-rate assumption.

Include AI subscription/API costs and estimated error-related costs
where reliable evidence exists.

Separate measured results from assumptions and projections.

## Research limitations
All three datasets are synthetic.
The experiments are controlled pilot tests, not evidence of
performance in an actual operating business.
"""

protocol_path = experiment_folder / "experimental_protocol.md"
protocol_path.write_text(protocol.strip(), encoding="utf-8")

#   Experiment A logging template
enquiry_log = pd.DataFrame(columns=[
    "task_id",
    "human_time_seconds",
    "ai_processing_seconds",
    "ai_review_seconds",
    "ai_total_time_seconds",
    "human_quality_score",
    "ai_quality_score",
    "ai_correction_count",
    "ai_unsupported_claims",
    "ai_escalation_correct",
    "notes"
])

#    Experiment B logging template
feedback_log = pd.DataFrame(columns=[
    "review_id",
    "true_sentiment",
    "predicted_sentiment",
    "true_topic",
    "predicted_topic",
    "processing_time_seconds"
])

#   Experimenting C logging template
product_log = pd.DataFrame(columns=[
    "product_id",
    "human_time_seconds",
    "ai_processing_seconds",
    "ai_review_seconds",
    "ai_total_time_seconds",
    "human_quality_score",
    "ai_quality_score",
    "ai_correction_count",
    "ai_unsupported_claims",
    "notes"
])

logs = {
    "experiment_A_enquiries_log.csv": enquiry_log,
    "experiment_B_feedback_log.csv": feedback_log,
    "experiment_C_products_log.csv": product_log
}

for filename, dataframe in logs.items():
    dataframe.to_csv(experiment_folder / filename, index=False)

print("EXPERIMENT WORKSPACE CREATED")
print("\nFiles:")

for path in sorted(experiment_folder.iterdir()):
    print(path.name)

print("\nExperiment protocol saved.")
print("Three experiment logging templates saved.")
```
